# **A. REPRODUCIBLE NORMALIZATION PROBLEM**

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
 x = np.mean(X)
 x
```

```python
s = np.std(X)
s
```

```python
X_normalized = (X-x)/s
X_normalized
```

```python
X_normalized_mean = np.mean(X_normalized)
X_normalized_mean
```

```python
X_normalized_std = np.std(X_normalized)
X_normalized_std
```

```python
print ('X:\n', X)
print ('X_normalized: \n', X_normalized)
print ('X_normalized_mean: ', X_normalized_mean)
print ('X_normalized_std:', X_normalized_std)
```

```python
np.save ("X-normalized.npy", X_normalized)
```

# **C. ABOVE-MEAN SQUARES PROBLEM**


```python
S = np.arange(1,37,)
S=S**2
S=np.resize(S,(6,6))
S


```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean= S[S>S_mean]
above_mean
```

```python
print ('S:\n', S)
print ('S_mean:', S_mean)
print ('above_mean:\n', above_mean)
print ('Number of elements above mean:', np.size(above_mean))
```

```python
np.save ('above_mean.npy', above_mean)
```
